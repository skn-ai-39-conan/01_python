# match ... case
- 값의 패턴에 따라 조건을 분기 처리하는 제어문이다.
- python 3.10+ 지원

```python
match 표현식:
    case 값1:
        실행문 
    case 값2:
        실행문
    case _:
        기본 실행문
```

````python
def print_drink_price(sel_num: int) -> None:

    match sel_num:
        case 1:
            print('콜라: 1000원')
        case 2:
            print('사이다 : 900원')

print_drink_price(1)
print_drink_price(2)
````

````python
# case_: 확인 -> 기본값(나머지)
def print_drink_price_v2(sel_num: int) -> None:

    match sel_num:
        case 1:
            print('콜라: 1000원')
        case 2:
            print('사이다 : 900원')
        case _:
            print('없는 번호 입력됨')

print_drink_price_v2(1)
print_drink_price_v2(2)
print_drink_price_v2(3)
print_drink_price_v2(999)
````

### 계산기 분기

1. 첫 번째 숫자 입력
2. 연산 기호 입력
3. 두 번째 숫자 입력
4. 결과 출력
