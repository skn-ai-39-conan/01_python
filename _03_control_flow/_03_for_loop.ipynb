# for 반복문
- 반복 횟수가 지정된 반복에 사용되는 구문
- python for문은 iterable 객체의 요소를 하나씩 꺼내 실행하는 구문
(iterable: 반복할 수 있는, 순회할 수 있는)
  - iterable 객체: list, tuple, set, dict, rangem, str 등...

# 1. 문자열, 리스트, 튜플 순회
- 시퀀스 타입 객체 (index가 존재한다)
- index 단위로 순회

````python
text: str = 'hello world'

for ch in text: # text(문자열에서 한 글자씩 꺼내 ch에 대입)
    print(ch)   # 모든 글자를 한번씩 꺼내면 반복 종료


print('종료됨')
````

````python
# 리스트 순회
lst: list[int] = [10, 20, 30, 40]

for num in lst:
    print(num)
````

# 2. dict, set 순회
- dict를 `for...in` 구문을 이용해서 순하면 key만 반환된다.
  - item 단위로 반환 받길 원하면 `dict.items()` 사용

- set은 순서가 없는 자료형이기 때문에 출력 순서는
  실행 환경에 따라 달라진다.

````python
# dict 순회
from typing import Any

# Any : 어떤 타입이든 작성 가능하다는 것을 나타내는 힌트용 타입(클래스)
user_info: dict[str, Any] = {
    'name': '홍길동',
    'age': 32,
    'hpbbies' : ['독서', '운동']
}

for key in user_info:
    print(f"{key} : {user_info[key]})")


# [dict item 순회]
print("--- item 순회 ---")
# for (key, value) in user_info.items():
for key, value in user_info.items(): # () 생략
    print(f'{key} : {value}')
````

````python
# set 순회
# - set 내부에 저장된 값을 꺼내는 유일한 방법
#   - set은 index가 없기 때문에 원하는 값만 꺼내는 것이 불가능
nums: set[int] = {1, 3, 4, 5, 6, 7, 8, 2, 4, 5, 14, 41}
for num in nums:
    print (num)
````

````python
# nums에 저장된 짝수만 찾아서 list에 옮겨담기
nums: set[int] = {1, 3, 4, 5, 6, 7, 8, 2, 4, 5, 14, 41}
even_list : list[int] = [] # 짝수만 저장할 리스트
odd_list: list[int] = []

for num in nums:
    if num % 2 == 0: # 2로 나눠서 나머지가 0인 수 == 짝수
        even_list.append(num) # num 옮겨닮기
    else:
        odd_list.append(num)

print("evenlist :", even_list)
print("odd_list :", odd_list)
````

## 3. range 객체

- 정수 범위를 표현하는 객체
- `range(start, end, step)`
  - start: 시작할 값
  - end: 끝 값 (미포함)
  - step: 증가 간격 (미작성 시 기본값1)


````python
rng = range(1, 5, 1) # 1 시작, 5 미만까지, 1씩 증가
                     # -> (1, 2, 3, 4)

print("rng :", rng, type(rng))

for num in rng:
    print("num :", num)
````

````python
# range를 이용한 0부터 10까지 추가되어있는 list 생성하기
nums: list[int] = list(range(0, 11))

print("nums :", nums)
````

````python
# 1부터 100까지의 모든 정수 합 구하기

n_sum = 0 # 누적 수행 시 결과에 영향을 미치지 않기 위해 0 사용

for num in range(1, 101, 1):
    # n_sum = n_sum + num
    n_sum += num # 복합 대입 연산자

print(n_sum) 
````

### (참고) 복합 대입 연산자
- 기존 변수를 이용한 연산 결과를
  다시 기존 변수에 대입하는 연산자

- +=, -=, *=, /=, //=, %=, **= ... 등이 존재

```python
# a에 20을 누적하는 방법
a = 10

# 방법 1
a = a + 20

# 방법 2 (복합 대입, 추천!)
a += 20

```

````python
a = 10
a += 5
print(a) # 15

b = 100
b -= 30
print(b) # 70

c = 4
c *= 5
print(c) # 20

d = 7
d /= 2
print(d) # 3.5
````

## 4. 중첩 반복문
- 반복문 안에 반복문이 존재하는 구조
- 다차원 구조를 생성하거나, 접근하는 용도로 많이 사용

시간 - 1년 1월 1일 0시 0분  0초

````python
# 중첩 반복문 예시
for a in range(1,4):
    for b in range(1,3):
        print(a, b)
````

````python
# 구구단 2단 출력하기
dan:int = 3 # 단 지정

for num in range(1,10,1):
   print(f"{dan} * {num} = {dan * num}")
````

````python
## 2단 부터 9단까지 출력하기
for dan in range(2, 10, 1):
    for num in range(1,10,1):
        print(f"{dan} * {num} = {dan * num}")
````

````python
# 시작 단과 끝 단을 입력 받아서 해당 범위 만큼만 구구단 출력학
start_dan = int(input(" 시작 단을 입력하세요. "))
end_dan = int(input(" 끝 단을 입력하세요. "))

for dan in range(start_dan, 10, 1):
    for num in range(1, end_dan + 1,1):
        print(f"{dan} x {num} = {dan * num}")
````

## 5. zip(), enumerate()

- zip(): 여러 iterable 객체의 요소를 같은 순서끼리 묶는다.
  - iterable 객체: str, list, set, dict, tuple ...

  
- enumerate(): 인덱스와 요소를 함께 반환하다.

````python
# zip() 객체 만들기 + 활용하기
names: list[str] = ['김영일', '박철수', '홍길동']
ages: list[int] = [20, 30, 40]

z = zip(names, ages) # 두 list의 같은 index 요소끼리 묶임

print(z, type(z)) # zip은 단독 출력 양식을 지원하지 X

# zip에서 요소를 하나씩 꺼냄
# -> 각 요소는 name, age 두 개의 값이 저장되어있다.
# --> 꺼내서 저장하는 for 변수도 두개를 묶은 tuple로 준비한다.
for (name, age) in z:
    print(name, age)
````

````python
# enumerate
foods: list[str] = ['떡볶이', '김밥', '족발', '삼겹살', '라면']

for (index, food) in enumerate(foods):
    print(f'{index}번 인덱스: {food}')
````

# 6. 반복문을 이용하여 개수, 합계, 평균 구하기

````python
scores: list[int] = [20, 43, 91, 79, 60]

# 전체 점수 합계:
# 전체 점수 평균:
# 60점 이상 점수의 개수:

score_sum = 0
cnt = 0 # 60점 이상 세는 변수

for score in scores:
    score_sum += score # scores 점수를 누적 == 합계 구하기
    if score >= 60:
        cnt += 1

# 평균 구하기 == 합계 / 요소 개수
scores_avg = score_sum / len(scores)


print('전체 점수 합계 :', score_sum)
print('전체 점수 평균 :', scores_avg)
print('60점 이상 점수 개수 :', cnt)
````

````python
# list[dict] 이용해서 점수 평균 구하기

from typing import Any

students: list[dict[str, Any]] = [
    {'name' : '홍길동', 'score': 100},
    {'name' : '김철수', 'score': 60},
    {'name' : '박미영', 'score': 90}
]

score_sum = 0
score_avg = 0

for student in students:
    score_sum += student.get('score')


print(score_sum)
print(score_sum / len(students))
````
