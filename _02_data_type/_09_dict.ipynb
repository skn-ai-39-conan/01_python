# dict
- 사전(dictionary) 자료형
- 하나의 요소(item)는 key:value 쌍으로 존재

- 같은 dict 내에서 key는 중복될 수 없다.
    - set은 순서 유지 X, dict는 가능(python 3.7+)

- value는 중복 가능하다

- key는 hashable 이어야 한다.
    - list, dict, set을 제외한 데이터 타입만 가능하다
    - 보통 str을 제일 많이 사용

- dict는 index 대신 key를 이용해서 value를 조회한다.

```python
# dict 생성
dct = {} # 빈 dict
print(type(dct))

dct ={
    "a" : 10,
    "b" : 20,
    "c" : 30
}

print(dct, type(dct))
```

```python
# tuple을 key로 지정
point_map = {(10, 20) : 'a 좌표'}
print(point_map, type(point_map))

# list를 key로 지정
#point_map2 = {[10, 20] : 'a 좌표'} # 에러 발생
```

## dict조회

```python
dct ={
    'a': 10,
    'b': 20,
    'c': 30
}
```

```python
# [key] 를 이용한 조회
print("dct['a'] :", dct['a'])
print("dct['b'] :", dct['b'])
print("dct['c'] :", dct['c'])
```

```python
# dict.get(key)를 이용한 조회
print("dct.get('a') :", dct.get('a'))
print("dct.get('b') :", dct.get('b'))
print("dct.get('c') :", dct.get('c'))
```

```python
# [], get() 차이

# [key] : key가 존재하지 않으면 KeyError 발생
print("dct['x'] :", dct['x'])
```

```python
# get() : key가 존재하지 않으면 None 반환
print("dct.get('x') :", dct.get('x'))
```

## dict 요소(item == K:V) 추가

```python
students = {
    "s001" : "홍길동",
    "s002" : "신사임당",
    "s003" : "이순신"
}

# students에 item 추가
students["s004"] = "강감찬"

# AI 답변(질문: dict.update() 메서드 사용 시 key를 따옴표로 감싸면 오류가 발생하는 이유)
# `students.update(s005="일론머스크")`는 `update()`에 **키워드 인자**를 전달하는 문법입니다. 키워드 인자의 이름은
# 변수명처럼 쓰므로 따옴표를 붙이지 않습니다. 이때 `s005`라는 이름이 문자열 `"s005"`인 딕셔너리 키로 추가됩니다.
#
# 따옴표를 붙여 `students.update("s005"="일론머스크")`라고 쓰면, 문자열은 키워드 인자의 이름으로 사용할 수 없어서 **문법
# 오류**가 발생합니다.
#
# 문자열 키를 명시하고 싶다면 딕셔너리로 전달하세요.
#
# students.update({"s005": "일론머스크"})
#
# 참고로 키워드 인자 방식은 키가 `s005`처럼 파이썬 식별자 규칙을 만족할 때만 사용할 수 있습니다.
students.update(s005 = "일론머스크") # K=V 형식으로 작성

# update() 수행 시 dict를 이용해서 여러 item을 한 번에 추가 가능
students.update({"s006" : "빌게이츠", "s007" : "백동현"})

print(students)
```

## dict 요소(item == K:V) 제거

```python
# pop(key)을 이용한 item 제거
removed = students.pop("s001") # value만 반환된다
print(removed)
print(students)
```

## dict 내장 함수

### dict api
- https://docs.python.org/ko/3.13/library/stdtypes.html#dict

```python
# dict() 생성 함수
dct2 = dict(a=10, b=20)
print(dct2)

dct3 = dict([('name', '신사임당'), ('age', 33)])
print(dct3)
```

```python
# dict.keys() : dict key만 추출
keys = students.keys()
print(keys, type(keys))
```

```python
# dict.values() : dict에서 values만 추출
values = students.values()
print(values, type(values))
```

```python
# dict.items() : dict의 item(K:V)를 tuple로 묶어 반환
items = students.items()
print(items, type(items))
```
