# Type Hint
- python은 동적 타이핑 언어
  -> 변수에 저장되는 값에 따라서 자료형이 결정된다

- type hint는 코드 작성 시 IDE(통합 개발 환경)과 정적 분석 도구
  데이터 타입을 추론하고 경고할 수 있도록 돕는 표시

- 중요한 점은 type hint 가 python 런타임에서 타입이 틀렸다고
에러를 발생시키지는 않는다 (경고만 한다!)

## 1. 동적 타이핑
- python에서는 같은 변수에 여러가지 서로 다른 타입의 변수를 대입할 수 있다.
- 대입되는 값에 따라서 자료형이 결정된다.

````python
x = '안녕'
print(x, type(x))

x = 100
print(x, type(x))
````

## 2. 변수의 타입 힌트
```python
변수명: 타입 = 값
```
- `: 타입` == 변수에 대입할 값을 타입 힌트

````python
greeting: str = 'hello'
print(greeting, type(greeting))

greeting = 12345 # 경고 발생, 하지만 정상 실행 가능
print(greeting, type(greeting))

num: int = 100
pi: float = 3.141592
is_true: bool = True
lst: list =[]
````

## 3. 컬렉션(자료구조 모음) 타입 힌트

- list, tuple, dict, set 같은 컬렉션에도 힌트 적용 가능
```python
변수명: 컬렉션[타입] = 값
```
- list\[int\]: 정수를 담는 리스트
  
- tuple\[int, str, str\]: 첫 번째는 정수, 두 번째와 세 번째는 문자열인 튜플
  
- dict\[str, str | int\]: key는 문자열, value는 문자열 또는 정수인 딕셔너리
  
- set\[str\]: 문자열을 담는 집합

````python
names: list[str] = ['홍길동', '고길동']

user: tuple[int, str, list[str]] = (10, "user1", [1, 2, 3])

# key는 str
# value는 str 또는 int 또는 float
info: dict[str, str | int | float] = {
    "a": "abc",
    "d": 10,
    "c": 3.14
}

chars: set[str] = {'a', 'b', 'c'}
````

## 4. 함수 타입 힌트
함수의 매개변수와 반환값에 타입 힌트를 지정하면 함수의 사용 방법을 명확하게 표현할 수 있습니다.

```python
def add(a: int, b: int) -> int:
    return a + b
```

````python
# 함수 선언 및 정의
def add(a: int, b: int) -> int:
    return a + b

result: int = add(3, 5)
print(result)
````

## 상수 표현과 Final

- 수학적 의미: 상수 == 항상 같은 수 == 변하지 않는 수
- 개발에서의 의미: 상수 == 저장된 값이 변하지 않는 변수

---
- python은 값을 변경할 수 없는 상수 문법이 별도로 존재하지 않음
- 관례상 대문자로 변수명이 작성된 경우를 `상수`로 취급
- typing.Final을 이용해서 힌트 제공 가능

````python
from typing import Final

# Final: 마지막 ==> 마지막으로 대입되는 값
MAX_COUNT: Final[int] = 100
print(MAX_COUNT)

MAX_COUNT = 200 # 실제로 재대입을 막을 순 없음
print(MAX_COUNT)
````
